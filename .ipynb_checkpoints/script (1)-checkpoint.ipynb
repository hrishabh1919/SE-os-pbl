```python
class SystemMonitorApp:
    """
    A real-time system monitoring application that displays CPU, memory,
    disk, and network usage using Tkinter and Matplotlib.
    """
    
    def __init__(self, root):
        # Setup main window
        self.root = root
        self.root.title("System Resource Monitor")
        self.root.geometry("900x700")
        
        # Initialize data storage
        self.timestamps = []
        self.cpu_data = []
        self.memory_data = []
        self.disk_data = []
        self.network_sent = []
        self.network_received = []
        
        # Maximum data points to store
        self.max_data_points = 60  # 1 minute of data
        
        # Set up UI components
        self.setup_ui()
        
        # Start the monitoring cycle
        self.update_stats()
```

```python
def update_stats(self):
    """Update system statistics and schedule the next update."""
    try:
        # Get current timestamp
        current_time = time.time()
        
        # Collect system metrics
        cpu_usage = psutil.cpu_percent(interval=0.1)
        memory_info = psutil.virtual_memory()
        
        # Determine appropriate disk path for the OS
        disk_path = "C:\\" if os.name == "nt" else "/"
        disk_usage = psutil.disk_usage(disk_path)
        
        # Get network I/O information
        net_io = psutil.net_io_counters()
        
        # Update data collections
        self.timestamps.append(current_time)
        self.cpu_data.append(cpu_usage)
        self.memory_data.append(memory_info.percent)
        self.disk_data.append(disk_usage.percent)
        self.network_sent.append(net_io.bytes_sent / (1024 * 1024))
        self.network_received.append(net_io.bytes_recv / (1024 * 1024))
        
        # Enforce data limits
        if len(self.timestamps) > self.max_data_points:
            self.timestamps = self.timestamps[-self.max_data_points:]
            self.cpu_data = self.cpu_data[-self.max_data_points:]
            self.memory_data = self.memory_data[-self.max_data_points:]
            self.disk_data = self.disk_data[-self.max_data_points:]
            self.network_sent = self.network_sent[-self.max_data_points:]
            self.network_received = self.network_received[-self.max_data_points:]
        
        # Update text labels and graphs
        self.update_labels(cpu_usage, memory_info, disk_usage, net_io)
        self.update_graphs()
        
        # Schedule next update
        self.root.after(1000, self.update_stats)
        
    except Exception as e:
        print(f"Error updating stats: {e}")
        self.root.after(1000, self.update_stats)
```

```python
def update_labels(self, cpu_usage, memory_info, disk_usage, net_io):
    """Update the text labels with current system information."""
    self.cpu_label.config(
        text=f"CPU Usage: {cpu_usage:.1f}%"
    )
    
    self.memory_label.config(
        text=f"Memory Usage: {memory_info.percent:.1f}% (Used: {memory_info.used / (1024**3):.2f} GB, "
             f"Total: {memory_info.total / (1024**3):.2f} GB)"
    )
    
    self.disk_label.config(
        text=f"Disk Usage: {disk_usage.percent:.1f}% (Used: {disk_usage.used / (1024**3):.2f} GB, "
             f"Total: {disk_usage.total / (1024**3):.2f} GB)"
    )
    
    self.network_label.config(
        text=f"Network: ↑ {self.network_sent[-1]:.2f} MB, ↓ {self.network_received[-1]:.2f} MB"
    )
```

```python
def update_graphs(self):
    """Update all four graphs with current data."""
    # Format time labels for x-axis
    time_labels = [time.strftime("%H:%M:%S", time.localtime(t)) for t in self.timestamps]
    
    # Clear and update CPU graph
    self.axs[0, 0].clear()
    self.axs[0, 0].plot(time_labels, self.cpu_data, color='#FF5733', 
                      linewidth=2, marker='o', markersize=3)
    self.axs[0, 0].set_title("CPU Usage", fontweight='bold')
    self.axs[0, 0].set_ylabel("Percent (%)")
    self.axs[0, 0].set_ylim(0, 100)
    self.axs[0, 0].grid(True, linestyle='--', alpha=0.7)
    
    # Format all x-axis labels
    for ax in self.axs.flat:
        # Only show a subset of timestamps to avoid overcrowding
        if len(time_labels) > 10:
            step = len(time_labels) // 5
            visible_indices = range(0, len(time_labels), step)
            visible_labels = [time_labels[i] if i in visible_indices else '' for i in range(len(time_labels))]
            ax.set_xticks(range(len(time_labels)))
            ax.set_xticklabels(visible_labels, rotation=45)
        else:
            ax.set_xticks(range(len(time_labels)))
            ax.set_xticklabels(time_labels, rotation=45)

        # Set up the graph colors and styles in setup_ui method
    self.graph_colors = {
        'cpu': '#FF5733',       # Bright red-orange
        'memory': '#33A1FF',    # Bright blue
        'disk': '#33FF57',      # Bright green
        'network_sent': '#9B59B6',    # Purple
        'network_received': '#F39C12'  # Orange
    }
    
    def main():
        """Main entry point for the application."""
        root = tk.Tk()
        app = SystemMonitorApp(root)
        root.mainloop()


    # Run application when script is executed directly
    if __name__ == "__main__":
        main()
```
